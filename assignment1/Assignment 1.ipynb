## Problem 1

Create a function `get_utility` that will read the Sushi Preference dataset and returns the utility matrix as a pandas Data Frame where the columns are the sushi names and the labels are the user IDs. Represent unrated items as `np.nan`. Sort by user ID.

```python
def get_utility():
    base = "/mnt/data/public/sushipref/sushi3-2016/"
    scores = pd.read_csv(base + "sushi3b.5000.10.score", sep=" ", header=None)
    users = pd.read_csv(base + "sushi3.udata", sep="\t", header=None)
    items = pd.read_csv(base + "sushi3.idata", sep="\t", header=None).sort_values(0)
    df = scores.replace(-1, np.nan)
    df.columns = items[1].to_numpy()
    df.index = users[0].to_numpy()
    return df.sort_index()
```

## Problem 5

Create a function `user_recommend` that accepts a user ID and utility matrix then returns the 10 best sushi recommendations for that `user` by a user-based collaborative filtering recommender system using (up to) $k=50$ nearest neighbors. Consider only neighbors with at least 3 commonly rated items as the `user`. Follow the preprocessing and conventions used in class. Sort by decreasing estimated rating then by ascending order of the sushi name.

```python
def user_recommend(user, df_utility, k=50, min_common=3):
    R = df_utility
    rated = R.notna().to_numpy()
    mu = R.mean(axis=1)                              # per-user mean
    C = R.sub(mu, axis=0)                            # mean-centred, NaN kept
    Z = C.fillna(0).to_numpy()
    pos = R.index.get_loc(user)

    norms = np.linalg.norm(Z, axis=1)
    with np.errstate(divide="ignore", invalid="ignore"):
        sims = (Z @ Z[pos]) / (norms * norms[pos])   # cosine on centred, zero-filled
    common = rated.astype(int) @ rated[pos].astype(int)

    ok = (common >= min_common) & np.isfinite(sims)
    ok[pos] = False
    cand = np.flatnonzero(ok)
    nbrs = cand[np.argsort(-sims[cand], kind="stable")[:k]]

    s = sims[nbrs]
    num = s @ Z[nbrs]                                # sum sim * centred rating
    den = np.abs(s) @ rated[nbrs].astype(float)      # sum |sim| over neighbours who rated item
    with np.errstate(divide="ignore", invalid="ignore"):
        pred = mu.iloc[pos] + num / den

    out = pd.Series(pred, index=R.columns)
    out = out[(~rated[pos]) & (den > 0)]
    out = out.reset_index()
    out.columns = ["name", "pred"]
    out = out.sort_values(["pred", "name"], ascending=[False, True])
    return out["name"].head(10).tolist()
```

## Problem 6

Create a function `item_recommend` that accepts a user ID and utility matrix then returns the 10 best sushi ID recommendations for that `user` by an item-based collaborative filtering recommender system using (up to) $k=10$ nearest neighbors. Consider only neighbors with at least 3 commonly rated items as the `item`. Follow the preprocessing and conventions used in class. Sort by decreasing estimated rating then by ascending order of the sushi name.

```python
def item_recommend(user, df_utility, k=50, min_common=3):
    R = df_utility
    rated = R.notna().to_numpy()
    mu = R.mean(axis=1)
    C = R.sub(mu, axis=0)                            # centre by user mean (adjusted cosine)
    Z = C.fillna(0).to_numpy()
    pos = R.index.get_loc(user)

    norms = np.linalg.norm(Z, axis=0)
    with np.errstate(divide="ignore", invalid="ignore"):
        S = (Z.T @ Z) / np.outer(norms, norms)       # item x item
    common = rated.T.astype(int) @ rated.astype(int)
    S = np.where((common >= min_common) & np.isfinite(S), S, np.nan)
    np.fill_diagonal(S, np.nan)

    mine = np.flatnonzero(rated[pos])                # items the user rated
    preds = {}
    for i in np.flatnonzero(~rated[pos]):
        s = S[i, mine]
        valid = ~np.isnan(s)
        if not valid.any():
            continue
        idx = np.flatnonzero(valid)
        top = idx[np.argsort(-s[idx], kind="stable")[:k]]
        w = s[top]
        den = np.abs(w).sum()
        if den == 0:
            continue
        preds[R.columns[i]] = mu.iloc[pos] + (w @ Z[pos, mine[top]]) / den

    ranked = sorted(preds.items(), key=lambda kv: (-kv[1], kv[0]))
    return [n for n, _ in ranked[:10]]
```

## Problem 7

Create a function `cgd_recommend` that accepts a user ID and utility matrix then returns the 10 best sushi recommendations for that `user` by a CGD-based recommender system using 20 latent components. Stop when the RMSE is at most 1.03. Sort the recommendations by decreasing estimated rating then by ascending order of the sushi name.

```python
def cgd_recommend(user, df_utility, n_comp=20, target=1.03,
                  lr=0.01, reg=0.0, seed=0, max_iter=5000):
    R = df_utility.to_numpy()
    M = ~np.isnan(R)
    R0 = np.where(M, R, 0.0)
    rng = np.random.default_rng(seed)
    P = rng.normal(0, 0.5, (R.shape[0], n_comp))
    Q = rng.normal(0, 0.5, (R.shape[1], n_comp))
    for _ in range(max_iter):
        E = M * (R0 - P @ Q.T)
        rmse = np.sqrt((E ** 2).sum() / M.sum())
        if rmse <= target:
            break
        P, Q = P + lr * (E @ Q - reg * P), Q + lr * (E.T @ P - reg * Q)
    pos = df_utility.index.get_loc(user)
    est = pd.Series(P[pos] @ Q.T, index=df_utility.columns)
    est = est[~M[pos]].reset_index()
    est.columns = ["name", "pred"]
    est = est.sort_values(["pred", "name"], ascending=[False, True])
    return est["name"].head(10).tolist()
```
