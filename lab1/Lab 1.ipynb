```python
import pandas as pd

R = "/mnt/data/public/bgg/bgg-19m-reviews.csv"
M = "/mnt/data/public/bgg/games_detailed_info.csv"

df = pd.read_csv(R, index_col=0, usecols=[0,1,2,4])
df = df[df["rating"] >= 1]
print(len(df), df["user"].nunique(), df["ID"].nunique())
print(df.duplicated(["user","ID"]).sum())

per_user = df.groupby("user").size()
print((per_user >= 20).sum())

meta = pd.read_csv(M)
print(meta["type"].value_counts())
meta = meta[meta["type"] == "boardgame"]
print(len(set(df["ID"]) - set(meta["id"])), df["ID"].nunique())

cnt = df.groupby("ID").size().rename("n_ratings")
g = meta.set_index("id")[["yearpublished"]].join(cnt, how="inner")
print(g[g["n_ratings"] >= 50].groupby("yearpublished").size().tail(12))
```

```python
import ast
def parse(x):
    try:
        return ast.literal_eval(x) if isinstance(x, str) else []
    except (ValueError, SyntaxError):
        return []
meta["mechanics"] = meta["boardgamemechanic"].apply(parse)
meta["categories"] = meta["boardgamecategory"].apply(parse)

dff = df[df["user"].isin(per_user[per_user >= 20].index)]
cnt = dff.groupby("ID").size().rename("n_ratings")
g = meta.set_index("id")[["yearpublished"]].join(cnt, how="inner")
print(((g["yearpublished"].between(2020, 2021)) & (g["n_ratings"] >= 50)).sum())
```

```python
def playable(meta, n):
    return meta[(meta["minplayers"] <= n) & (meta["maxplayers"] >= n)]["id"]

# check for bad values first
print((meta["minplayers"] == 0).sum(), (meta["maxplayers"] == 0).sum())
print(meta[["minplayers", "maxplayers"]].isna().sum())

bad = (meta["minplayers"] == 0) | (meta["maxplayers"] == 0)
print(bad.sum(), bad.mean())
```
